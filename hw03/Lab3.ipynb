### Original assignment
> Write a suitable CFG grammar, able to correctly parse all sentences, with all their senses. Use as inspiration the grammars discussed in class.

### Goal and approach
Define CFG non-terminals, productions and vocabulary covering the three sentences. Ambiguity is represented by multiple possible derivations; the grammar is a small demonstration grammar, not a complete model of English. Some syntactic readings may be overgenerated, so inspect whether each tree corresponds to a plausible sense.

```python
from nltk import CFG, ChartParser
```

```python
grammar = CFG.fromstring("""
    S -> NP VP

    NP -> Det N | N | Adj NP | NP Conj NP | NP PP | GerundP
    GerundP -> Ving NP

    VP -> Modal VP | Cop Adj | Aux Ving Adv | V NP | VP PP
    PP -> Prep NP

    Det -> 'the'
    N -> 'professors' | 'men' | 'women' | 'reporter' | 'actor' | 'camera' | 'entrance'
    Adj -> 'visiting' | 'old' | 'annoying'
    Ving -> 'visiting' | 'waiting'
    Modal -> 'can'
    Cop -> 'be'
    Aux -> 'were'
    Adv -> 'outside'
    V -> 'interviewed'
    Conj -> 'and'
    Prep -> 'with' | 'near'
""")

print(grammar)
```

```python
sentences = [
    "Visiting professors can be annoying.",
    "Old men and women were waiting outside.",
    "The reporter interviewed the actor with the camera near the entrance."
]

parser = ChartParser(grammar)

for sentence in sentences:
    words = sentence.lower().replace(".", "").split()
    trees = list(parser.parse(words))

    print(sentence)
    print("Number of interpretations:", len(trees))
    print()
```

## Task 4 — Probabilistic CFG to Chomsky Normal Form (0.30p)

### Original assignment
> Write a program which transforms a Context Free Grammar, as the one discussed in class, into a Chomsky Normal Form grammar, also converting the probabilities. Test it on the grammar from the course.

### Goal and approach
The statement refers to converting **probabilities**, so the implementation uses an NLTK **PCFG** rather than a probability-free CFG. For an epsilon-free PCFG, the algorithm replaces terminals in long rules, binarizes productions, eliminates unary nonterminal rules and preserves the aggregate probability of each string.

**Limitation:** The grammar from the course is not included in the provided PDF or notebook. The test grammar below is **illustrative, not the course grammar**. To satisfy the exact final test requirement, substitute the grammar supplied in class.

### 4.1 Conversion algorithm

CNF only permits productions of the form `A -> B C` or `A -> 'word'` (apart from the special empty-start case). The implementation below handles **epsilon-free PCFGs**, including cycles of unary nonterminal rules when their total weight is finite. It uses three steps:

1. Replace terminals in longer rules with fresh preterminal symbols of probability 1.
2. Convert rules with more than two symbols to binary rules. The original probability is attached to the first binary rule; other introduced rules have probability 1.
3. Remove unary nonterminal productions by computing their weighted transitive closure, preserving the total probability of every terminal string.

Different derivation trees of the original grammar can collapse into the same CNF production. Such production probabilities are **summed**. This implementation explicitly rejects epsilon productions rather than silently altering their probabilities. It also checks that each LHS's outgoing probabilities sum to 1.

```python
from collections import defaultdict
from itertools import count
import numpy as np
from nltk import Nonterminal, PCFG, ProbabilisticProduction


def to_cnf_pcfg(grammar, tolerance=1e-9):
    """Convert an epsilon-free NLTK PCFG to probability-preserving CNF.

    The probabilities of complete strings are preserved, including contributions
    from distinct unary derivations. Epsilon productions are not supported.
    """
    if not isinstance(grammar, PCFG):
        raise TypeError("Expected an nltk.PCFG with production probabilities")
    if any(len(p.rhs()) == 0 for p in grammar.productions()):
        raise ValueError("Epsilon productions require a separate nullable-rule transformation")

    original = grammar.productions()
    names = {str(s) for p in original for s in (p.lhs(), *p.rhs()) if isinstance(s, Nonterminal)}
    serial = count()

    def fresh(prefix):
        while True:
            name = f"__CNF_{prefix}_{next(serial)}"
            if name not in names:
                names.add(name)
                return Nonterminal(name)

    # Store probabilities in a dictionary to merge duplicate productions.
    rules = defaultdict(float)
    terminal_symbols = {}

    for production in original:
        lhs, rhs, probability = production.lhs(), production.rhs(), production.prob()
        if len(rhs) >= 2:
            new_rhs = []
            for symbol in rhs:
                if isinstance(symbol, Nonterminal):
                    new_rhs.append(symbol)
                else:
                    if symbol not in terminal_symbols:
                        terminal_symbols[symbol] = fresh("TERM")
                    new_rhs.append(terminal_symbols[symbol])
            rhs = tuple(new_rhs)

        if len(rhs) <= 2:
            rules[(lhs, rhs)] += probability
        else:
            current = lhs
            for offset in range(len(rhs) - 2):
                next_nt = fresh("BIN")
                rules[(current, (rhs[offset], next_nt))] += probability if offset == 0 else 1.0
                current = next_nt
            rules[(current, tuple(rhs[-2:]))] += 1.0

    for terminal, symbol in terminal_symbols.items():
        rules[(symbol, (terminal,))] += 1.0

    nts = sorted({lhs for lhs, _ in rules} |
                 {s for _, rhs in rules for s in rhs if isinstance(s, Nonterminal)}, key=str)
    indices = {symbol: i for i, symbol in enumerate(nts)}
    unary = np.zeros((len(nts), len(nts)))
    non_unary = []
    for (lhs, rhs), probability in rules.items():
        if len(rhs) == 1 and isinstance(rhs[0], Nonterminal):
            unary[indices[lhs], indices[rhs[0]]] += probability
        else:
            non_unary.append((lhs, rhs, probability))

    # (I - U)^(-1)[A, B] sums probabilities over all unary paths A =>* B.
    try:
        closure = np.linalg.solve(np.eye(len(nts)) - unary, np.eye(len(nts)))
    except np.linalg.LinAlgError as exc:
        raise ValueError("Cannot eliminate a non-convergent unary cycle") from exc
    if np.min(closure) < -tolerance:
        raise ValueError("Unary cycles do not define convergent nonnegative path weights")

    converted = defaultdict(float)
    for lhs, rhs, probability in non_unary:
        col = indices[lhs]
        for target, row in indices.items():
            value = closure[row, col] * probability
            if value > tolerance:
                converted[(target, rhs)] += value

    products = [ProbabilisticProduction(lhs, rhs, prob=value)
                for (lhs, rhs), value in converted.items() if value > tolerance]
    result = PCFG(grammar.start(), products)
    if not result.is_chomsky_normal_form():
        raise AssertionError("Conversion did not produce CNF")
    by_lhs = defaultdict(float)
    for prod in result.productions():
        by_lhs[prod.lhs()] += prod.prob()
    for lhs, total in by_lhs.items():
        if abs(total - 1.0) > 1e-7:
            raise ValueError(f"Probabilities for {lhs} sum to {total}, not 1")
    return result
```

### 4.2 Example grammar and probability comparison

```python
from nltk import InsideChartParser

example_pcfg = PCFG.fromstring("""
S -> NP VP [1.0]
NP -> Det N [0.6] | Name [0.4]
Det -> 'the' [1.0]
N -> 'professor' [0.5] | 'reporter' [0.5]
Name -> 'alice' [1.0]
VP -> V NP PP [0.4] | V NP [0.6]
V -> 'saw' [1.0]
PP -> P NP [1.0]
P -> 'with' [1.0]
""")

cnf_pcfg = to_cnf_pcfg(example_pcfg)
print("Converted grammar is CNF:", cnf_pcfg.is_chomsky_normal_form())
print(cnf_pcfg)

# Inside probability sums over all derivations; it is not a Viterbi maximum.
def sentence_probability(grammar, words):
    from nltk import InsideChartParser
    return sum(tree.prob() for tree in InsideChartParser(grammar).parse(words))

examples = [
    "alice saw the reporter".split(),
    "the professor saw alice".split(),
    "alice saw the reporter with the professor".split(),
]
for words in examples:
    original_prob = sentence_probability(example_pcfg, words)
    cnf_prob = sentence_probability(cnf_pcfg, words)
    assert np.isclose(original_prob, cnf_prob, rtol=1e-9, atol=1e-12)
    print(" ".join(words), ":", original_prob, "->", cnf_prob)
```

4.3 Automated unit tests
The tests verify CNF structure, preserved sentence probabilities, mixed terminals, unary rules, ambiguous derivations, and rejection of unsupported epsilon productions.

```python
import unittest


class TestPCFGtoCNF(unittest.TestCase):
    def assert_same_probability(self, source, samples):
        converted = to_cnf_pcfg(source)
        self.assertTrue(converted.is_chomsky_normal_form())
        for sentence in samples:
            self.assertAlmostEqual(sentence_probability(source, sentence.split()),
                                   sentence_probability(converted, sentence.split()), places=9)

    def test_course_substitute(self):
        self.assert_same_probability(example_pcfg,
            ["alice saw the reporter", "the professor saw alice",
             "alice saw the reporter with the professor"])

    def test_mixed_terminals_and_chain(self):
        source = PCFG.fromstring("""
        S -> A [1.0]
        A -> B [0.6] | 'a' B 'c' [0.4]
        B -> 'b' [1.0]
        """)
        self.assert_same_probability(source, ["b", "a b c"])

    def test_convergent_unary_cycle(self):
        source = PCFG.fromstring("""
        S -> A [1.0]
        A -> A [0.2] | 'a' [0.8]
        """)
        converted = to_cnf_pcfg(source)
        self.assertTrue(converted.is_chomsky_normal_form())
        # The original cyclic parse forest is infinite; verify its exact series.
        self.assertAlmostEqual(sentence_probability(converted, ["a"]), 1.0)

    def test_ambiguous_rules(self):
        source = PCFG.fromstring("""
        S -> A [0.5] | B [0.5]
        A -> 'x' [1.0]
        B -> 'x' [1.0]
        """)
        self.assert_same_probability(source, ["x"])

    def test_reject_epsilon(self):
        source = PCFG.fromstring("""
        S -> A [1.0]
        A -> [0.5] | 'a' [0.5]
        """)
        with self.assertRaises(ValueError):
            to_cnf_pcfg(source)


suite = unittest.defaultTestLoader.loadTestsFromTestCase(TestPCFGtoCNF)
result = unittest.TextTestRunner(verbosity=2).run(suite)
assert result.wasSuccessful()
```

## Task 5 — Practical Application (0.15p)

### Original assignment
> Describe an application n which constituency and/or dependency parsing can provide useful information. Explain briefly why. (1/2 pag.)

### Answer
One useful application of constituency and dependency parsing is **information extraction from news articles**. News reports often describe events by identifying an actor, an action, and the person or object affected by that action. Dependency parsing can help a system identify these roles even when the sentence contains additional descriptions or when word order changes.

For example, in *“The reporter interviewed the actor with the camera near the entrance”*, the dependency relations can identify *reporter* as the subject of *interviewed* and *actor* as its object. The prepositional phrases *with the camera* and *near the entrance* add information, but their attachment is ambiguous. Depending on the interpretation, the camera may be an instrument used in the interview or an object associated with the actor, while *near the entrance* may describe the camera, the actor, or the location of the interview. A single statistical parser output is only one possible analysis, not proof of the intended meaning.

Constituency parsing contributes complementary information. By identifying noun phrases, verb phrases, and prepositional phrases, it allows a system to extract complete entities such as *the actor with the camera* instead of collecting isolated nouns. It also helps identify alternative phrase groupings when a sentence has multiple valid readings.

A practical pipeline could tokenize a news article, generate dependency and constituency analyses, extract subject–verb–object relations, and store them in a structured database. Such records could then support semantic search, question answering, or automatic event summaries. Parsing alone cannot guarantee factual correctness or resolve every ambiguity, but it provides useful syntactic evidence that makes downstream extraction more systematic and interpretable.
